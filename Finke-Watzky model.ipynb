```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ---------------------------------
# Parameters
# ---------------------------------
A0 = 1.0
k1 = 0.1
k2 = 1.0

t_max = 10
dt = 0.01
t = np.arange(0, t_max, dt)


# ----------------------------------
# Numerical solution (Euler method)
# ----------------------------------
B_num = np.zeros_like(t)
A_num = np.zeros_like(t)

B = 0.0
A = A0

for i in range (len(t)):
    B_num[i] = B
    A_num[i] = A

    dB = (k1 + k2 + B) * (A0 - B)
    dA = -dB

    B += dB * dt
    A += dA * dt


# ----------------------------------
# Analytical solution
# ----------------------------------
exp_term = np.exp(-(k1 + k2 * A0) * t)

B_analytical = (A0 * (1 - exp_term)) / (1 + (k2 * A0 / k1) * exp_term)



# ----------------------------------
# Plots
# ----------------------------------
plt.figure()
plt.plot(t, B_num, label = "Numerical B(t)")
plt.plot(t, B_analytical, "--", label = "Analytical B(t)")
plt.xlabel("Time")
plt.ylabel("B(t)")
plt.legend()
plt.title("Finke-Watzky model: Numerical vs Analytical")
plt.show()
```

```python
# The numerical solution of the Finke–Watzky model shows excellent agreement with the analytical solution, validating
# the implementation. The system exhibits sigmoidal growth due to the transition from nucleation-dominated to
# autocatalytic growth.
```
